# Plotly Express (PX)

Plotly Express (PX) es la __interfaz de alto nivel__ de la biblioteca Plotly para Python. Es ideal para crear\
__gráficos interactivos__ de forma rápida y sencilla gracias a su API concisa y potente.

🧩 ¿Qué es Plotly Express?

* Es parte del módulo plotly.express, recomendado como punto de partida para generar la mayoría de los gráficos 

* Cada función de PX, como px.line(), px.bar(), px.scatter(), etc., construye un plotly.graph_objects.Figure\
internamente, pero con mucho menos código — de 5 a 100 veces más compacto que si usaras Graph Objects directamente 

* Cuenta con más de 30 tipos de visualización: gráficas de líneas, barras, tortas, mapas, 3D, distribución, etc.

✅ Beneficios de usar Plotly Express

1. Código breve y claro:

    fig = px.line(df, x='fecha', y='ventas', color='país', markers=True)

    fig.show()

    — esta línea genera un gráfico interactivo completo, elegante y funcional 

3. Interactividad por defecto:

    * Tooltip al pasar el ratón.

    * Zoom y paneo en tiempo real.

    * Posibilidad de ocultar/mostrar series clicando en la leyenda 
   
4. Estilos y defaults inteligentes:

    * Elige paletas de colores, ejes, títulos y dimensiones por defecto.

    * Permite sobrescribir fácilmente esos estilos si querés más control .

5. Integración con datos diversos:

    * Funciona con DataFrames (tidy/wide), diccionarios y arrays de NumPy.

    * También acepta Pandas, Polars, xarray, GeoPandas, etc. 

6. Modificable si necesitás más control:

    * Devuelve una Figure que podés editar con métodos como update_layout, add_trace, etc. 

Mas info en: plotly.com

```python
import pandas as pd
```

```python
# reding the csv file
df_population_raw = pd.read_csv('population_total.csv')
```

```python
df_population_raw.head(
```

```python
# dropping null values
df_population_raw.dropna(inplace=True)

# making pivot table
df_pivot = df_population_raw.pivot(index='year', columns='country', 
                                    values='population')

# Selecting some countries
df_pivot = df_pivot[['United States','India', 'China', 'Indonesia', 'Brazil']]

# Showing pivot table
df_pivot
```

# 1. Interactive Lineplot

So we will be able to see the data behind the lines.

```python
import plotly.express as px

fig = px.line(df_pivot, 
              x=df_pivot.index, 
              y=df_pivot.columns,
              markers=True,
              title="Gráfico interactivo")
fig.update_layout(yaxis_title="Population")
fig.show()
```

```python
# So this graph allows us not only to see the plot but we can see the data behind
# and we get more detail about this lineplot
```

# 2. Interactive Barplot

```python
# Selecting only one year (2020)
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]

# transpose dataframe from rows to columns or viceversa
df_pivot_2020 = df_pivot_2020.T
```

```python
df_pivot_2020
```

```python
fig = px.bar(
    df_pivot_2020,
    x=df_pivot_2020.index,  # categorías en el eje X
    y='2020.0',               # valores
    labels={'x': 'País', '2020': 'Población 2020'},
    title='Población por país en 2020',
    text='2020'            # muestra el valor sobre cada barra
)
fig.update_layout(xaxis_title='País', yaxis_title='Población 2020')
fig.show()
```

## 3.2 Interactive Barplot grouped by "n" variables

```python
# Selecting a few years
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020])]
df_pivot_sample
```

```python
# DataFrame en formato largo
df_long = df_pivot_sample.reset_index().melt(
    id_vars='year', 
    var_name='country', 
    value_name='population'
)

fig = px.bar(
    df_long,
    x='year',
    y='population',
    color='country',         # Una barra por país
    barmode='group',         # Barras una al lado de otra
    title='Población por país a lo largo de los años',
    labels={'population': 'Población', 'year': 'Año', 'country': 'País'}
)

fig.update_layout(xaxis_title='Año', yaxis_title='Población')
fig.show()
```

```python
# So here we have this multiple bar plots grouped by 5 variables.. and here we can
# see in detail the population of each country in the year we want.. So for example,
# India in 2020 had a population of 1.38B of inhabitants.. US in 1990 had a population
# of 252M inhabitants.
```

# 3. Interactive Piechart

```python
# Reseteando el índice para convertir los países en columna
df_plot = df_pivot_2020.reset_index()

# Renombrando la columna de población 
df_plot = df_plot.rename(columns={2020.0: 'population'})

# Creando el gráfico de torta
fig = px.pie(
    df_plot,
    names='country',
    values='population',
    title='Proporción de población por país en 2020',
)

# Mostrar porcentaje y nombre dentro del gráfico
fig.update_traces(textinfo='percent+label')

fig.show()
```

# 4. Interactive Boxplot

```python
# convertiendo la Serie en un DataFrame
df_us = df_pivot['United States'].reset_index()
df_us.columns = ['year', 'population']

# creando el boxplot interactivo
fig = px.box(
    df_us,
    y='population',
    points='all',  # muestra todos los puntos individuales
    title='Boxplot de la población de Estados Unidos',
    labels={'population': 'Población'},
)

fig.show()
```

```python
# Reorganizando el DataFrame de formato ancho a largo
df_long = df_pivot.reset_index().melt(id_vars='year', 
                                      var_name='country', 
                                      value_name='population')

# Creando el boxplot
fig = px.box(
    df_long,
    x='country',
    y='population',
    points='all',  # muestra puntos individuales (opcional)
    title='Distribución de la población por país',
    labels={'country': 'País', 'population': 'Población'}
)

fig.show()
```

# 5. Interactive histogram

```python
# Convirtiendo la Serie en DataFrame
df_us = df_pivot[['United States']].reset_index()

# Creando histograma
fig = px.histogram(
    df_us,
    x='United States',
    nbins=10,  # se puede ajustar la cantidad de bins
    title='Histograma de la población de Estados Unidos',
    labels={'United States': 'Población'}
)

fig.show()
```

```python
# the first interval has 1 value, the 2nd and 3rd interval have 2 values, ..., the last interval has 6 values
```

```python
# Convirtiendo de formato ancho a largo
df_hist = df_pivot[['United States', 'Indonesia']].reset_index().melt(
    id_vars='year',
    var_name='country',
    value_name='population'
)

# Creando histogramas interactivos con colores por país
fig = px.histogram(
    df_hist,
    x='population',
    color='country',
    nbins=15,
    barmode='overlay',  # usa 'group' si querés verlos lado a lado
    title='Histograma de población: Estados Unidos vs Indonesia',
    labels={'population': 'Población'}
)

fig.show()
```

```python
# se puede clickear en la leyenda el pais que queramos y solo mostrará el histograma asociado a ese pais.
```

# 6. ScatterPlot

```python
df_pivot.head()
```

```python
# df_pivot tiene años como índice y países como columnas
df_long = df_pivot.reset_index().melt(id_vars='year', var_name='País', value_name='Población')

# Creando el scatter plot
fig = px.scatter(
    df_long,
    x='year',
    y='Población',
    color='País',
    title='Población por país a lo largo del tiempo (Scatter Plot)',
    labels={'year': 'Año'}
)

fig.show()
```
