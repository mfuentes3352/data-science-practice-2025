```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'],
                    'job': ['Doctor', 'Statistician', 
                            'Engineer', 'Developer']})
```

## 1.1 Inner Join

The Inner Join returns matching values between two tables. In this case, it will be the matching\
data between two dataframes. So if we have dataframe 1 on the left and dataframe 2 on the right, the\
data that they have in common is the data that will be in the middle (the intersection of the two sets).

And to write the code that produces this inner join, we have to use the merge() method:

df1.merge(df2, on='id', how='inner')

id is the column that both dataframes have in common, and there will be elements in common in this id column.

```python
df_inner_join = df1.merge(df2, on='id', how='inner')
df_inner_join
```

```python
# Solo aparecen esas filas porque df1 y df2 tienen en comun la columna id
# y porque en esa columna id tienen en comun los elementos C y D, esos son los
# elementos que están en la intersección de los dos conjuntos. Entonces junta
# las filas de ambas tablas y las pone una al lado de la otra, solo las filas 
# que tienen elementos en comun, en este caso C y D.
```

```python
# so we get a df that consists of these three columns and this data that corresponds to
# the C and D id.
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df_inner_join.shape)
```

```python
# The first two dfs have the same number of rows and columns. But the last df, so
# the df that represents the iner join has only two rows and three columns.. So
# this happens because these two dataframes have two rows in common, which are in the
# C and D rows. And here the three columns are the id, the job and the age column.
```

```python
# And that's how the inner join can be made using the merge() method.
```

# Example

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv', low_memory=False)
```

```python
# selecting the columns
df_movies = df_movies[['imdb_title_id', 'title', 'year', 
                       'genre', 'country']]
df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

```python
# merging df_movies and df_ratings (inner join)
df_movies.merge(df_ratings, on='imdb_title_id', how='inner')    
# default value of the parameter how is inner
# pd.merge(df_movies, df_ratings, on='imdb_title_id') tambien funcionoa
```

```python
# and now we can see that this df have the columns of df_movies: 'imdb_title_id',# 'title', 'year', 'genre', 'country'
# and 'total_votes', 'mean_vote' belong to df_ratings

# only the common 'imdb_title_id' are in this df
```

```python
# And that's how an inner join between df_movies and df_ratings can be done.
```
