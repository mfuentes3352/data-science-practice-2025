```python
import pandas as pd
```

# Concatenate Horizontally

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'job': ['Doctor', 'Statistician', 
                            'Engineer', 'Developer']})
```

```python
pd.concat([df1, df2], axis=1)
```

```python
# And we get a single datafreme with the three columns.
# So we see the id, the age and the job in this single df
```

# Example

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv', low_memory=False)
```

```python
# selecting the columns
df_movies = df_movies[['imdb_title_id', 'title', 'year', 
                       'genre', 'country']]
df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

```python
# shape of dataframes that we'll concatenate
print(df_movies.shape)
print(df_ratings.shape)
```

```python
# concatenating df_movies and df_ratings on 'imdb_title_id' (horizontally)
df_movies.set_index('imdb_title_id', inplace=True)
df_movies.head()
```

```python
df_ratings.set_index('imdb_title_id', inplace=True)
df_ratings.head()
```

```python
df = pd.concat([df_movies, df_ratings], axis=1)
df
```

```python
# and now we have this concatenation. So this df has the four columns of the df_movies
# and the two columns of the df_ratings, and here the index is the id column
```

```python
# And that's how two dataframes can be concatenated horizontally.
```
