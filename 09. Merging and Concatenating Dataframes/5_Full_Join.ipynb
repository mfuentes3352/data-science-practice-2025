## 1.2 Full Joins

Full joins allows us to join all the elements between two tables. So if we have a dataframe 1 on the left\
and a dataframe 2 on the right, the data that belong to both dfs will belong to the full join (the union of the two sets).

All the elements of the tables are taken by the full join because this full join takes all the elements\
in this two dataframes.

We have to find a column in common, a column that both dataframes have. And to make this full join we have\
to use the merge() method:

df1.merge(df2, on='id', how='outer')

on the "on" parameter we set it equal to the column that these two dfs have in common.\
in the "how" parameter we set it equal to outer, because this is a full join also known as an outer join.

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'],
                    'job': ['Doctor', 'Statistician', 
                            'Engineer', 'Developer']})
```

```python
df1
```

```python
df2
```

```python
df1.merge(df2, on='id', how='outer')
```

```python
# and now this produces the dataframe we want it. And here we see that the
# A and B id have null values (because df1 doesn't have the 'job' column) and
# the E and F id have null values (because df2 doesn't have the 'age' column)
# While the C and D id don't have null values because these are the elements in
# common between df1 and df2.
```

```python
# shape
print(df1.shape)
print(df2.shape)
print(df1.merge(df2, on='id', how='outer').shape)
```

## Example

```python
# merging df_movies and df_ratings (outer join)
```

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv', low_memory=False)
```

```python
# selecting the columns
df_movies = df_movies[['imdb_title_id', 'title', 'year', 
                       'genre', 'country']]
df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
# both dfs have in common the column imdb_title_id 
```

```python
df_movies.merge(df_ratings, on='imdb_title_id', how='outer')    
```

```python
# and we get this dataframe that is the full join of the df_movies and the df_ratings..
# As we can see, these columns: 'title,'year','genre','country' belong to df_movies
# and these: 'total_votes','mean_vote', belong to df_ratings.
```

```python
# And that's how full join between df_movies and df_ratings can be done.
```
