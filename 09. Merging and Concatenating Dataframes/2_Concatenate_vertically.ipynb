# concat()

In Pandas, we can use the concat() method to concatenate two dataframes. And there are two different\
ways to concatenate dataframes. We can concatenate dataframes vertically and horizontally.

axis = 0 means vertical concatenation (concatenating along the rows),\
axis = 1 means horizontal concatenation (concatenating along the columns)

pd.concat([df1, df2]), axis = 0 by default

# Concatenate vertically

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['E', 'F', 'G', 'H'],
                    'age': [40, 21, 19, 24]})
```

```python
pd.concat([df1, df2], axis=0, ignore_index=True)
```

```python
# and this is how the resulting df looks, df1 is above df2
```

```python
# Given that we set ignore_index=True, we'll see we get 0, 1, 2, 3, ..., 7
# in the indexes. And we see we don't get the 0, 1, 2, 3 indexes repeated
# because we ignored the original indexes. So this df will have a new index
# ignoring the indexes from df1 and df2.
```

# Example

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
```

```python
# selecting the columns
df_movies = df_movies[['imdb_title_id', 'title', 'year', 
                       'genre', 'country']]
```

```python
# extracting a 50% sample of the original dataframe (df_movies)
df_movies2 = df_movies.sample(frac=0.5, random_state=42)
```

```python
# shape of the dataframes that we'll concatenate
df_movies.shape, df_movies2.shape
```

```python
# concatenating df_movies and df_sample (vertically along the rows)
df = pd.concat([df_movies, df_movies2], axis=0, ignore_index=True)
df
```
