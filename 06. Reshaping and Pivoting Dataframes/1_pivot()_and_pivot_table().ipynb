## pivot()

This method reshapes data based on column values (it does not support data aggregation).

```python
# reading and showing the dataset
df_gdp = pd.read_csv('gdp.csv')
df_gdp
```

```python
# This df shows how the GDP evolves over the years for each country
```

In this example we're going to obtain a different view of the original data set to see\
the evolution of the GDP per capita over the years for each country.

```python
# reshaping the dataframe with .pivot()
df_gdp.pivot(index='year', columns='country', values='gdppc')

# Returns reshaped DataFrame organized by given index / column values.

# Reshapes data (produce a "pivot" table) based on column values. Uses
# unique values from specified `index` / `columns` to form axes of the
# resulting DataFrame. This function does not support data
# aggregation, multiple values will result in a MultiIndex in the
# columns.

# quisiera que el dataframe tenga otra forma, por ejemplo que en las filas
# tenga cierta variable como indice (p. ej, el año), y en las columnas aparezcan otras variables
# (p. ej los paises, que cada pais se una columna independiente (y que no se repitan))
# y que en las celdas no aparezcan esos datos sino otros, (p. ej valores de pbi per capita)..
# entonces deberiamos cambiar la forma del df, seleccionar estas variables y, a su vez, seleccionar
# con que valores voy a rellenar las celdas de datos.
```

```python
# So here, as we can see, we have the new view of this dataframe. And it looks much better, it's more readable..
# because we can see the GDP evolution over the years for each country..
```

## .pivot_table()

It creates a spreadsheet-style pivot table (and it supports data aggregation).

```python
# reading and showing the dataset
df_sales = pd.read_excel('supermarket_sales.xlsx')    
df_sales

# antes de ejecutar estas lineas, abrir una nueva ventana de Anaconda Prompt,
# activar el entorno virtual en el que estamos trabajando y ejecutar pip install openpyxl
```

```python
# and here we can see we have different columns about what a specific person bought in a supermarket
# and here we have the 'Branch', the 'City', the 'Gender' and different data.
```

```python
# making a pivot table and adding an aggregate function

# Filtrar solo columnas numéricas
numeric_cols = df_sales.select_dtypes(include='number').columns

# Crear la pivot table solo con esas columnas
df_sales.pivot_table(index="Gender", values=numeric_cols, aggfunc="sum")
```

```python
# making a pivot table, adding an aggregate function and selecting some columns
df_sales.pivot_table(index="Gender", 
                     values=["Quantity", "Total"], 
                     aggfunc="sum")
```

```python
# so far we know that female spend 167882.925 and male spend 155083.824 in this supermarket 

# but with pivot tables, we can even know in which product lines this money is spent.
```

```python
# making a pivot table that says how much male and female spend in each Product Line
df_sales.pivot_table(index="Gender",
                     columns="Product line",
                     values="Total", 
                     aggfunc="sum")
```

```python
# and now as we can see here, how much female and male spend in each product line..

# So we can quickly see, for example that male spend slightly more on Electronic accessories than female,
# female spend more money on Fashion accessories than male, and that kind of makes sense, and also in sports
# female spend more money than male..

# So we could easily see all of that by using the pivot_table() method in Pandas,
```
