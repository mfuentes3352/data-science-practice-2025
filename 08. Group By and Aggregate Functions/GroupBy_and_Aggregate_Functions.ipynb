# 1. agg()

```python
# calculating the sum of sales in thousands
df_car_sales['Sales_in_thousands'].agg('sum')
```

```python
# calculating the mean of numeric columns
df_car_sales.select_dtypes(include='number').agg('mean').round(2)
```

```python
# calculating the mean and count values
df_car_sales.select_dtypes(include='number').agg(['mean', 'count']).round(2)
```

```python
# Applying different aggregation fuctions to a column:
# calculating the sum and mean of 'Sales_in_thousands'
# and sum and max of 'Price_in_thousands'

df_car_sales.agg({'Sales_in_thousands':['sum', 'mean'],
                  'Price_in_thousands':['sum', 'max']}).round(2)
```

```python
# applying aggregate functions over the columns. observation:
# df_car_sales[['Sales_in_thousands', 'Price_in_thousands']].agg('sum', axis=1) (NaN is considered 0)
# is equivalent to calculate the sum df_car_sales['Sales_in_thousands'] + df_car_sales['Price_in_thousands']
# (if there is a NaN value, the result of the operation is NaN)
```

```python
# aggregate different functions and rename the index of the resulting dataframe

df_car_sales.agg( x = ('Sales_in_thousands', 'sum'),
                   y = ('Price_in_thousands', 'sum')).round(2)
```

# 2. The Split-Apply-Combine Strategy

```python
df_car_sales['Vehicle_type'].value_counts()
```

```python
# splitting data into separate groups

passenger_filter = df_car_sales['Vehicle_type'] == 'Passenger'
car_filter = df_car_sales['Vehicle_type'] == 'Car'
# we got boolean values in this series
```

```python
# applying an operation
passenger_average = df_car_sales[passenger_filter]['Sales_in_thousands'].mean().round(2)
car_average = df_car_sales[car_filter]['Sales_in_thousands'].mean().round(2)
```

```python
# applying df_car_sales[passenger_filter] we only got the observations whose 'Vehicle_type' is Passenger
# then with the operation we obtain the average 'Sales_in_thousands' but only for the passenger Vehicle_type
```

```python
# combining results
pd.DataFrame({'Vehicle_type':['Car', 'Passenger'],
              'Sales_in_thousands':[car_average, passenger_average]})
```

# 4. groupby() and agg()

```python
# finding the minimum and maximum values on each column
df_car_sales.groupby('Vehicle_type').min()
```

```python
df_car_sales.groupby('Vehicle_type').max()
```

```python
df_car_sales.groupby('Vehicle_type').agg(['min','max'])
```

```python
# setting a name for aggregated columns
# finding the min Engine_size and max Horsepower
df_car_sales.groupby('Vehicle_type').agg( min_engine_size = ('Engine_size', 'min'),
                                          max_horsepower = ('Horsepower', 'max') )
```

```python
# so here we got the Vehicle_type in the index, because we grouped by Vehicle_type..
# and then we have here two columns, so 1st is the min_engine_size and this column has the 
# minimum value (2.0 and 1.0) and the 2nd column is the max_horsepower that has the max values
```

```python
# calculating the sum of Sales_in_thousands and mean of Price_in_thousands
df_car_sales.groupby('Manufacturer').agg( sum_sales = ('Sales_in_thousands', 'sum'),
                                          mean_price = ('Price_in_thousands', 'mean') )
```

```python
# and here we have a dataframe, and in this df we grouped the data by Manufacturer
# and then we sum the values in the sum_sales column and then we calculated the mean
# in the mean_price column.. So here we have all the values for all the manufacturers.
```

```python
# these operations are equivalent to using the functions

# df_car_sales.groupby('Manufcaturer').sum()['Sales_in_thousands']
# df_car_sales.groupby('Manufcaturer').mean()['Price_in_thousands']

# but with the aggregate functions we can group by a categorical variables, excecute
# multiple operations with a single instruction and get a dataframe as a result ... 
```

```python
# And that's how the groupby() and agg() methods can be used together.
```

# 5. Lambda

We're going to learn how to apply our own function to our group by object using the lambda function.

```python
# getting actual sales and price values
df_car_sales.groupby('Manufacturer').sum().apply(lambda x: x*1000)[['Sales_in_thousands','Price_in_thousands']]
```

```python
# subtract the values from the mean of each group
numerical_cols = df_car_sales.select_dtypes(include='number').columns

df_car_sales.groupby('Manufacturer')[numerical_cols].apply(lambda x: x-x.mean())

# we calculate the mean of each category, for example the mean Price_in_thousands of the group Audi, 
# and then we subtract the mean to each value in the Price_in_thousands, for example.
```

```python
# and as we can see, this lambda function that we created shows us the values that are under the mean
# and over the mean. So we can recofnize which are good values.
```

```python
# When aggrgate functions are applied inside the lambda function (for example, the mean)
# we apply this function to the groups that we get after using the group by method. 

# In this case, we grouped by 'Manufacturer' with the groupby() method, we calculated the
# mean over the Manufacturer groups, and we subtract the mean from the values..
```

```python
# And that's how the lambda function works and how complex functions can be created using the apply method and the lambda function.
```

# 6. The filter() method

The filter() method helps us filter values based on an aggregate function when we use it with the groupby() method.

```python
# calculating the mean of Sales_in_thousands
df_car_sales[numerical_cols].mean()
```

```python
# filtering certain values out based on sum() and returning a subset
def filter_func(x):
    '''it filters groups based on 
    the sum of the sales in thousands'''
    return x['Sales_in_thousands'].sum() > 52    # we're going to filter based on this condition
# x represents a dataframe
    
df_filter = df_car_sales.groupby(['Manufacturer']).filter(filter_func)
df_filter 
# So here in my filter method we specify that we want to apply that function
```

```python
# all the values that are in this df passed the filter. So this means that the sum of the Sales
# (in each group) is greater than 52. For example, the sum of the sales in the Audi group is > 52, and so on.
```

```python
# if we group by manufacturer, we filter the numerical cols and sum the sales in thousands
df_car_sales.groupby(['Manufacturer'])[numerical_cols].sum()['Sales_in_thousands'].sort_values().head(6)
```

```python
# and sort them ascending, we obtain that the first 6 manufacturer's Sales are < 52.. so they shouldn't
# be in the previous df because they were filtered out. To verify this we can do the following:
df_car_sales[df_car_sales['Manufacturer'].isin(['Porsche','Jaguar','Saab','Infiniti','Audi','BMW'])].shape
```

```python
# the shape of this df is 13 rows * 7 columns. and now let's compare the shape of the original df
# and the shape of the df_filter, to see how many rows we lost in the filter.
df_car_sales.shape
```

```python
# and here we see that in the original df we had 157 rows
df_filter.shape
```
